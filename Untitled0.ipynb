```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
dataset = pd.read_csv("math_science_scores_30_observations.csv")

# Display dataset information
dataset.info()

# Set plot style
sns.set(style="whitegrid")

# Create figure
plt.figure(figsize=(8, 6))

# Create scatter plot
sns.scatterplot(
    x="math score",
    y="science score",
    hue="Gender",
    style="Branch",
    s=100,
    data=dataset
)

# Add title and labels
plt.title("Math Score vs Science Score", fontsize=16)
plt.xlabel("Math Score", fontsize=14)
plt.ylabel("Science Score", fontsize=14)

# Add legend
plt.legend(title="Gender")

# Display plot
plt.show()
```
